```python
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
```

```python
X,y = make_classification(n_samples=10000,n_features=10,n_informative=3)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2,random_state=23)
```

```python
X_train
```

```python
bg = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=.5,
    bootstrap=True,# with reolacement
    random_state=42
)
```

```python
bg.fit(X_train,y_train)
```

```python
accuracy_score(y_test,bg.predict(X_test))
```

```python

bag = BaggingClassifier(
    estimator=SVC(),
    n_estimators=500,
    max_samples=0.25,
    bootstrap=True,
    random_state=42
)
     
```

```python
bag.fit(X_train,y_train)
```

```python
accuracy_score(y_test,bag.predict(X_test))
```

Pasting 

```python
bag_pas = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=.5,
    bootstrap=False,# no reolacement
    random_state=42
)
```

```python
bag_pas.fit(X_train,y_train)
```

```python
accuracy_score(y_test,bag_pas.predict(X_test))
```

Random Subspaces

```python

bag_random_sub = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=1.0,
    bootstrap=False,
    max_features=0.5,
    bootstrap_features=True,
    random_state=42
)
```

```python
bag_random_sub.fit(X_train,y_train)
```

```python
accuracy_score(y_test,bag_random_sub.predict(X_test))
```

Random Patches

```python
bag_random_patch = BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=.5,
    max_features=.5,
    bootstrap=True,
    bootstrap_features=True,
    random_state=43
)
```

```python
bag_random_patch.fit(X_train,y_train)
```

```python
accuracy_score(y_test,bag_random_patch.predict(X_test))
```

OOB (OUT OF BAG SAMPLES)

* Samples that are never seen by algorithm due to bootstrapping

```python

bag_oob= BaggingClassifier(
    estimator=DecisionTreeClassifier(),
    n_estimators=500,
    max_samples=0.25,
    bootstrap=True,
    oob_score=True,
    random_state=42
)
```

```python
bag_oob.fit(X_train,y_train)
```

```python
bag_oob.oob_score_
```

* Bagging generally gives better results than Pasting
* Good results come around the 25% to 50% row sampling mark
* Random patches and subspaces should be used while dealing with high dimensional data
* To find the correct hyperparameter values we can do GridSearchCV/RandomSearchCV

# 1. Body Mass Index(BMI)

# BMI = Weight (kg) / (Height (m)²)

### BMI is a widely used indicator of overall body composition and health.

````python
df['BMI'] = df['Weight_kg']/(df['Height_cm']/100)**2

df.head()
````

```Interpretation:

BMI < 18.5: Underweight

18.5 ≤ BMI < 24.9: Normal weight

25 ≤ BMI < 29.9: Overweight

BMI ≥ 30: Obese

# 2. Waist-to-Height Ratio (WHtR)

## WHtR = Waist Circumference (cm) / Height (cm)

### WHtR is another measure of abdominal obesity and is a better predictor of cardiovascular risk than BMI in some cases.

````python
df['WHtR'] = df['Waist_Circumference_cm']/df['Height_cm']
df.head()
````

### WHtR > 0.5 indicates increased risk of cardiovascular diseases.

# 3. Lean Body Mass(LBM) Ratio

## LBM Ratio = Lean Body Mass / Total Body Weight

### Shows proportion of muscle & organs

````python
body_fat_weight = df['Weight_kg'] - ((df['Weight_kg']*df['Body_Fat_Percentage'])/100)
lean_body_mass = df['Weight_kg'] - body_fat_weight
df['LBM_ratio'] = lean_body_mass/df['Weight_kg']
df.head()
````

# 4. Fat-Free Mass Index (FFMI)

# FMI = (Weight * (1 - Body Fat Percentage)) / (Height²)

### Similar to BMI but adjusted for lean mass


````python
df['FMI'] = (df['Weight_kg'] * (1 - (df['Body_Fat_Percentage'])/100)) / ((df['Height_cm']/100)**2)
df.head()
````

# 5. Visceral Fat Estimate

# Visceral Fat Estimate=Waist Circumference (cm)×Body Fat Percentage

### Visceral fat (fat around internal organs) is a strong predictor of cardiovascular disease. While direct measurement requires specialized equipment, you can estimate it using waist circumference and body fat percentage.



````python
df['vis_fat'] = df['Waist_Circumference_cm']*df['Body_Fat_Percentage']/100
df.head()
````

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/Users/shivendragupta/Desktop/Machine Learning/Feature Engineering/credit_data.csv')
```

```python
df.head()
```

```python
df = df.drop(columns=['Debt_to_Credit_Ratio','Late_Payment_Ratio','Recent_Credit_Opening_Ratio','Delinquency_Rate'])
```

# 1. Debt-to-credit Ratio

## This measures the amount of credit you're using compared to your total available credit limit.

 ## Debt-credit-ratio = Total Account Balance / Total Credit Limit.

```python
df['Debt_cred_ratio'] = df['Total_Account_Balance']/df['Total_Credit_Limit']
df.head()
```

### High ratio indicates high reliance of credit limit and a risky behavior

# 2. Late Payment Ratio

## This measures the frequency of late payments relative to the total number of payments due.



## Late Payment Ratio = Late Payments / Total Payments.

```python
df['late_payment_ratio'] = df['Late_Payments'] / df['Total_Payments']
df.head()
```

### A higher ratio indicates that more late payments that indicates irresponsible or risky behaviour 


# 3. Recent Credit Opening Ratio


## Recent_credit_opening_ratio = Accounts Opened Last 12 Months / Total Accounts

## This measures the proportion of credit accounts opended in last 12 months relative to total credit accounts




```python
df['Recent_credit_opening_ratio'] = df['Accounts_Opened_Last_12_Months']/df['Total_Accounts']
df.head()
```

### Higher value indicates that over reliance on credit indicating financial instability

# 4. Delinquency Rate

## Deliquency_rate  = Accounts 30+ Days Late / Total Accounts.

## This measures the ratio of accounts with late payment past 30 days to the total accounts

```python
df['Deliquency_rate'] = df['Accounts_30Plus_Days_Late']/df['Total_Accounts']
df.head()
```

### A higher delinquency rate indicate higher chances of payment defaults in future

# 5. Credit Health Score

## Credit Health Score=Credit Utilization Ratio+Late Payment Ratio+Delinquency Rate

```python
df['Credit_health_score'] = df['Debt_cred_ratio'] + df['late_payment_ratio'] + df['Deliquency_rate']
df.head()
```

### Lower the value better is the credit worthiness 

# 6. Average Balance per Account

## Indicates amount of account balance per account

## Total_Account_Balance / Total_Accounts

```python
df['Average_balance_per_account'] = df['Total_Account_Balance']/df['Total_Accounts']
df.head()
```

### A higher average account balance indicated higher reliance on credit 

# 7. Average Credit Limit per Account

## This measures the average credit utilization per account.

## Total_Credit_Limit / Total_Accounts

```python
df['Average_credit_limit_per_account'] = df['Total_Credit_Limit']/df['Total_Accounts']
df.head()
```

### Higher the average means higher the credit Worthiness

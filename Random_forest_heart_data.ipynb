```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier,GradientBoostingClassifier
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('/Users/shivendragupta/Desktop/Machine Learning/heart.csv')
df.head()
```

```python
from sklearn.model_selection import train_test_split

X = df.iloc[:,0:13]
y = df.iloc[:,-1]

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=.2,random_state=42)
X_train
```

```python
rf = RandomForestClassifier(n_estimators=1000)
```

```python
rf.fit(X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score

accuracy_score(y_test,rf.predict(X_test))
```

```python
gbc = GradientBoostingClassifier()
svc = SVC()
lr = LogisticRegression()
```

```python
gbc.fit(X_train,y_train)
svc.fit(X_train,y_train)
lr.fit(X_train,y_train)
```

```python
print(accuracy_score(y_test,gbc.predict(X_test)))
print(accuracy_score(y_test,svc.predict(X_test)))
print(accuracy_score(y_test,lr.predict(X_test)))
```

```python
rf_tuned = RandomForestClassifier(max_samples=.75,random_state=42)
```

```python
rf_tuned.fit(X_train,y_train)
```

```python
accuracy_score(y_test,rf_tuned.predict(X_test))
```

Hyper Parameter Tuning Using GridSearch cv

```python
from sklearn.model_selection import GridSearchCV
```

```python
n_estimator = [20,60,100,150,200,300,400,600,700,800,1000]
max_depth = [1,2,3,4,5,6,7,8,9,None]
max_samples = [.25,.5,.75,1]
bootstrap = [True,False]
min_sample_split = [1,2,3,4,5,6]
```

```python
param_grid = {'n_estimators': n_estimator,
               'max_depth': max_depth,
              'max_samples':max_samples,
              'bootstrap':bootstrap,
              'min_samples_split':min_sample_split,
             }
print(param_grid)
```

```python

from sklearn.model_selection import GridSearchCV

rf_grid = GridSearchCV(estimator = rf, 
                       param_grid = param_grid, 
                       cv = 5, 
                       verbose=2, 
                       n_jobs = -1)
```

```python
rf_grid.fit(X_train,y_train)
```

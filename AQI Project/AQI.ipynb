```python
df_c['Date'] = pd.to_datetime(df_c['Date'],format = '%d/%m/%Y')
# df_c['hour_of_day'] = df_c['Date'].dt.hour  # Hour of the day
df_c['day_of_week'] = df_c['Date'].dt.dayofweek  # 0 = Monday, 6 = Sunday
df_c['month'] = df_c['Date'].dt.month  # Extract month

# 10/03/2004	18.00.00
df_c
```

```python
def get_season(month):
    if month in [12, 1, 2]:
        return "Winter"
    elif month in [3, 4, 5]:
        return "Spring"
    elif month in [6, 7, 8]:
        return "Summer"
    else:
        return "Fall"

df_c['season'] = df_c['month'].apply(get_season)
```

```python
df_c['is_weekend'] = df_c['day_of_week'].isin([5, 6]) 
df_c
```

```python
df['Time'] = df['Time'].str.replace('.', ':', regex=False) 
df_c['time_of_day'] = pd.to_datetime(df['Time'], format='%H:%M:%S')
df_c
```

```python
def get_time_of_day(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"
df_c['hour_of_day'] = df_c['time_of_day'].dt.hour  
df_c['time_of_day_cat'] = df_c['hour_of_day'].apply(get_time_of_day)  # Assign category
df_c
```

# Simulated Annealing Algorithm

### Overview
Simulated Annealing is a probabilistic optimization algorithm inspired by the annealing process in metallurgy. It is used to find the global minimum of a loss function by allowing occasional uphill moves to escape local minima.

### Algorithm Steps
1. **Initialize the parameters** $ ( \theta) $ and set an initial temperature \( T \).
2. **For each iteration:**
   - Generate a new candidate solution $ ( \theta') $ by perturbing $ ( \theta) $
   - Compute the change in loss:
   $$
   \Delta J = J(\theta') - J(\theta)
   $$

   - If $ ( \Delta J < 0 )$, accept $ ( \theta) $
   - If $( \Delta J \geq 0 )$, accept $ ( \theta') $ with probability:
   $$
   P = \exp\left(-\frac{\Delta J}{T}\right)
   $$
   - Reduce the temperature \( T \) over time.
3. **Repeat until convergence.**

### Advantages
- **Can escape local minima**: By allowing occasional uphill moves, the algorithm can explore a wider solution space and potentially find the global minimum.
- **Suitable for non-convex optimization problems**: Simulated Annealing works well for complex, non-convex problems where other optimization techniques might get stuck in local minima.

### Disadvantages
- **Requires careful tuning of the cooling schedule**: The temperature cooling rate and stopping criteria are critical and need to be tuned for the problem at hand.
- **Computationally expensive**: Due to the probabilistic nature of the algorithm and the need to evaluate many candidate solutions, Simulated Annealing can be computationally intensive, especially for large problems.

```python
from sklearn.datasets import load_diabetes
import numpy as np
import pandas as pd
import random
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
# Simulated Annealing Algorithm
def simulated_annealing(loss_function, w, T0, alpha, T_min, max_iter):
    w = w  # Initial solution
    T = T0  # Initial temperature
    best_w = w
    best_f = loss_function(w)
    
    for k in range(max_iter):
        # Generate a new candidate solution by adding random noise
        w_new = w + np.random.normal(0, 1, len(w))  # Random perturbation
        
        # Compute the change in the objective function
        delta_f = loss_function(w_new) - loss_function(w)
        
        # Acceptance probability
        if delta_f < 0 or np.random.rand() < np.exp(-delta_f / T):
            w = w_new  # Accept the new solution
        
        # Update the best solution found so far
        if loss_function(w) < best_f:
            best_w = w
            best_f = loss_function(w)
        
        # Cool down the temperature
        T = alpha * T
        if T < T_min:
            break
    
    return best_w, best_f

# Define the loss function (Mean Squared Error)
def loss_function(w):
    # Add a column of ones to X for the bias term
    X_new = np.c_[np.ones(X.shape[0]), X]
    
    # Compute predictions: y_hat = X_new @ w
    y_hat = X_new @ w
    
    # Compute Mean Squared Error (MSE)
    mse = np.mean((y - y_hat) ** 2)
    return mse


np.random.seed(42)

w = np.ones(X.shape[1] + 1)  # Number of features + 1 for bias

# Run simulated annealing
T0 = 10  # Initial temperature
alpha = 1  # Cooling rate
T_min = 1e-19  # Minimum temperature
max_iter = 70000  # Maximum iterations

best_w, best_f = simulated_annealing(loss_function, w, T0, alpha, T_min, max_iter)
print("Best weights (including bias):", best_w)
print("Best MSE:", best_f)
```
